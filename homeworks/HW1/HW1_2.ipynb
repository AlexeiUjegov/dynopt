### Солверы

Все солверы, которые мы рассматриваем, очень похожи между собой, поэтому для экспериментов полезно использовать подходящую классовую структуру, выделив весь  базовый функционал в базовый класс.

Солвер -- это итеративная процедура, которая делает по одному шагу по дискретной сетке $t_i,~ i=0,..,N$, на каждом шаге вычисляя решение по предыдущему. Такие солверы нахываются одношаговыми, но многоэтапными (если нужно сделать несколько промежуточных вызовов правой части $f$). Таким образом, каждый солвер из рассматриваемых нами должен:
1. Делать шаг вычисления.
2. Делать последовательно $N$ шагов вычисления приближённого решения $\tilde{X}_{t_i}$ по значению $\tilde{X}_{t_{i-1}}$.
Шаг вычисления зависит от конкретного метода и поэтому его можно вынести в дочерний класс.

### Leap Frog

Этот метод специально сделан для решения систем уравнений вида

$$
\ddot{x} = f(x).
$$

Сначала система сводится к двум блокам уравнений:

$$
\dot{x} = v, ~~ \dot{v} = f(x).
$$

Переменные $x$ называются состояними, а $v$ -- импульсами (или скоростями). Как можно догадаться по названиям, этот метод придуман в первую очередь для гамильтоновых систем.

Схема метода

* Задаём шаг по времени $h>0$ (в этом варианте метода его ещё можно делать зависимым от времени)
* $v_{h/2} = v_0 + \frac{h}{2} f (x_0,v_0)$

* $~~~~v_{i+1/2} = v_{i-1/2} + f(x_i,v_i) h$

## Исследуем точность на примерах

Асимптотику метода можно оценить эмпирически, анализируя динамику ошибки. Рассмотрим один конкретный метод X.

Идея в следующем:

1. Задаётся последовательность сеток дискретизации $\Tau_1,..,\Tau_N$, каждая $\Tau_i=\left\lbrace t_1^i,...,t_{n_1}^i\right\rbrace$. Каждая следующая сетка должна быть более плотной, чем предыдущая.
2. Для каждой сетки дискретизации $\Tau_i$ вычисляется решение дифференциального уравнения методом Х.
3. При наличии точного решения (наш случай!) сравниваем численное решение $\tilde{X}(t^i_j)$ с истинным $X(t^i_j)$, например, так:

$$
E_i = \max_{j} \vert \tilde{X}(t^i_j) - X(t^i_j)\vert.
$$

4. По собранным $h_i = \max_j (t^i_{j+1}-t^i_j)$ и $E_i$ строится линейная регрессия (спойлер), вычисляя оценку наименьших квадаратов по схеме

$$
\ln E_i \approx \alpha \ln h_i + \beta. 
$$

Оценённый $\hat{\alpha}$ позволяет судить о динамике ошибки порядка $h^{\hat{\alpha}}$, а число $\alpha$ называется порядком точности метода. Чем больше порядок точности, тем меньше шагов нужно методу, чтобы достигнуть нужного уровня ошибки.
