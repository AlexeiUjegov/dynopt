Задача мененджмента -- понять, как оптимально пользоваться этим ресурсом, чтобы избежать негативных последствий.

В самом простом случае мы рассматриваем рост популяции рыбы $x(t)$ (измеряется в тоннах) согласно

$$
\dot{x} = rx - u , ~~x(0)=x_0,
$$

где $u$ -- это разрешаемая квота вылова (тонн/год), про которую известно, что $u \in [0,u_0]$. По итогам мы хотели бы вывести рыбу на заданный уровень численности $x(T)=x_T$ (что уже может быть априори недостижимо, но без этого задача становится плохо поставленной).

Мы хотели бы эксплуатировать ресурс, продавая рыбу на рынок, получая за неё суммарно (без учёта дисконтирования, в тысячах рублей)

$$
\int_0^T p u(t) dt,
$$

где $p$ -- заданная константа.

Ваша задача -- изучить, как управленческие решения по выдаче квот меняют систему и понять, как их выдавать оптимально.

## Линейный регулятор с квадратичным критерием (LQR) (50 баллов)

Это очень классическая задача и пример одного из первых значительных достижений теории управления. Несмотря на её простоту, любое усложнение может даваться непросто математически и технически, что демонстрируется общим количеством статей, вышедших на эту тему. 

Помимо этого, есть аналоги этой задачи для стохастических дифференциальных уравнений, а также в случае стохастической динамики в дискретном времени, где LQR решается в два этапа: строится алгоритм фильтрации данных на основе фильтра Калмана, а затем используются результаты для детерминированного управления. Одним из первых, кто стал решать эту задачу в стохастической постановке (правда в дискретном времени), стал Ричард Беллман, используя недавно изобретённый аппарат уравнений в частных производных, а также Рудольф Калман, который придумал алгоритм оптимальной фильтрации [в дискретном времени](https://www.unitedthc.com/DSP/Kalman1960.pdf) (1960), его коллега Ричард Бюси помог построить аналог [в непрерывном времени](https://people.duke.edu/~hpgavin/SystemID/References/KalmanBucy-ASME-JBE-1961.pdf). Задачами фильтрации активно занимались в советской школе, мотивируя разработку фильтрации прежде как инструмент для стохастических задач оптимального управления, значимой фигурой здесь выступал академик [Руслан Стратонович](https://www.phys.msu.ru/upload/iblock/914/RLS-book.pdf), решавший задачу нелинейной фильтарции раньше Калмана, который знал о работах Стратоновича. Они впервые встретились на первом когрессе по оптимальному управлению IFAC, который проходил в Москве в 1960 году.

### Постановка задачи и построение управления (10 баллов)

Известно, что система удовлетворяет линейному дифференциальному уравнению и линейно управляется:

$$
\dot{x} = A(t)x + B(t)u,
$$

$A(t) \in \mathbb{R}^{d \times d}, B(t) \in \mathbb{R}^{d \times k}$. Всё зависит от времени, мы это будем предполагать, но часто индекс $t$ будет опускаться.

Дано начальное состояние $x(0)=x_0$, фиксированы ограничения на $u(t) \in U$, но пока мы их опустим и положим $u(t) \in \mathbb{R}^k$, потому что их учёт концептуально достаточно простой, но нам важно не сомневаться, что мы можем всё и решение существует. В противном случае логично возможен $x(0)$ такой, что с таким слабым управлением задачу не решить, разрешимость задачи для любых начальных данных называется свойством _управляемости_ .

Цель: стабилизировать $x$ около точки $\overline{x} \in \mathbb{R}^d$ минимальными усилиями. Это достигается через решение оптимизационной задачи на фиксированном горизонте $T>0$

$$
\min_u \int_0^T (x-\overline{x})^T Q (x-\overline{x}) + u^TRu ~dt,
$$

где $Q,R$ -- заданные квадратные симметричные и положительно определённые матрицы, определяющие евклидову метрику для приближения $x$ и цену за использование управления $u$.

Для простоты мы полагаем $B$ и $Q,R$ константами. От этого отказаться непросто, но и не особо нужно: есть немногие задачи, где критерий меняется со временем (например [вот](https://pmc.ncbi.nlm.nih.gov/articles/PMC3991650/pdf/pone.0094925.pdf) ), но $B$ как правило константа, потому что она отражает способность управления воздействовать на среду .

Есть несколько способов решить эту задачу, мы зайдём со стороны принципа максимума. Для начала сделаем предварительные упрощения.

#### Предварительные упрощения

Прежде всего, обычная задача LQR -- про стабилизацию около нуля, в нашем же случае мы хотим стабилизироваться около $\overline{x}$. Чтобы это учесть, мы можем перейти к новым координатам $z = x -\overline{x}$. Можно решать задачу в старой постановке, но для этого нужно аккуратно сделать замены $x,u$, а также адаптировав технику ниже.

(Если вы приложите полное математическое решение, которое полностью корректно решает задачу в оригинальной постановке со всеми нюансами, можно получить +5 балла, это та же техника)

Для большего упрощения будем рассматривать

$\dot{z}=Az + Bu,$

полагая, что функционал записан для отклонений.

Итоговый функционал после сведения задачи к задаче на максимум:

$
\mathrm{max_{u}} -\int_0^T z^TQz + u^TRu ~dt.
$

Теперь визуально стало сильно проще. Кажется, мы ничего не сделали, но мы ещё увидим эффект при составлении уравнений.

### Имплементация (20 баллов)

Теперь мы готовы к самому интересному: как это всё посчитать?

Метод состоит в следующем.

Даны $x(0), \overline{x}, A(t), B(t), T>0$.

1. Решаем уравнение на $C(t)$ (численно, так как аналитически не всегда можно).
2. Синтезируем управление как $u= C(t) z(t)$ (формулы нашли выше).
3. Решаем уравнение на $z(t) $ с данным управлением И начальным условием $z(0)=x(0)-\overline{x}$.
4. Делаем обратную замену $x(t) = \overline{x} + z(t)$.

Мы всё умеем, кроме первого этапа, но он делается тоже несложно.

Пусть дано уравнение типа

$ \dot{C} + f(t,C(t)) = 0,~ С(T)=C_T$

Сделаем замену времени $s=T-t$, тогда $C(t)=C(T-s)=\tilde{C}(s), ~f(t,C(t))=f(T-s,C(T-s))=\tilde{f}(s,\tilde{C}(s))$ И производная $\dot{\tilde{C}}=d\tilde{C}/ds=-dC/dt$. Итого:

$$
-\frac{d\tilde{C}}{ds} + f(T-s,\tilde{C}(s)) = 0.
$$

Замена несложная, но теперь конечное условие стало начальным $\tilde{C}(0)=C_T$.

Выпишите уравнение Риккати после замены времени.
