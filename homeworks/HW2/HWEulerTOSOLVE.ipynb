## Физический маятник: не всё так очевидно (30 баллов)

Мы уже видели модель физического маятника на неделях дифференциальных уравнений. Теперь мы имеем возможность вывести эту модель методами вариационного исчисления. Вот общая методология, как происходит вывод уравнений в механике для систем с сохранением энергии (они называются консервативными). 

1. Записывается лагранжиан системы, в физике (конкретно в механике) этим словом называют разность (!) суммарной кинетической и потенциальной энергии системы:
$$
L(t, x, \dot{x}) = \frac{m \Vert \dot{x} \Vert_2^2 }{2} - U(x).
$$
Обычно эта функция определяется из самых базовых принципов (см. например[ Ландау и Лифшиц, Том 1 Механика, Глава 1](http://www.phys.nsu.ru/serebr/wp-content/uploads/2020/03/LL1.pdf)) и её записать относительно несложно.

2. Механическая система (она очень умная) живёт так, чтобы минимизировать некоторый функциионал за время существования $[0,T]$, конкретно
$$
\min_{x,\dot{x} \in M} \int_{0}^T L(t,x,\dot{x}) dt.
$$
Этот интеграл называется *действием*, а принцип минимизации называется *принципом наименьшего действия* или *принципом Гамильтона*. Для неконсервативных систем (где есть потери или добавления энергии) существует адаптированный подход, который называется *принципом Мопертюи*. Если присмотреться, то уравнение Эйлера -- это в точности то, что называется вторым законом Ньютона, а сила $F$ строится как градиент потенциальной энергии.

3. Нельзя забывать, что есть ограничения, они и дают специфику каждой задаче механики.

У такого вариационного подхода по сравнению с относительно школьным методом выписывания сил есть одно мощное преимущество, из-за которого инженеры-механики буквально живут этим методом: выписать энергию в сложных системах (например, механизмах для роботов) и получить уравнения прямой кинематики оказывается сильно проще, чем думать о том, какие где возникают силы и получать уравнения сразу из второго закона Ньютона.

### Вывод уравнений (10 баллов)

Маятник задаётся длиной стержня $L$, полагается, что стержень не имеет массы. Маятник -- это тело предельно малого размера (материальная точка) с массой $m$.

Положение маятника в декартовой системе задаётся как $x(t),y(t)$, точка $(0,0)$ находится в точке подвеса.

В итоге действие задаётся как

$$
\int_0^T m \frac{\dot{x}^2+\dot{y}^2}{2} - mg y ~dt.
$$

При этом есть ограничение: $x(t)^2 + y(t)^2 = L^2$, -- оно задаёт геометрию движения. В более сложных случаях даётся система таких ограничений, которые тоже выписываются обычно несложно.

Уравнение Эйлера в данной задаче -- это и есть дифференциальное уравнение физического маятника.

## Учёт ограничений: цепная линия (50 баллов)

Положим, мы закрепили цепь или нить в двух точках, не растягивая. Какую форму примет цепь в положении равновесия? Принцип наименьшего действия говорит, что такую, которая минимизирует общую потенциальную энергию. Решение такой (похожей) задачи позволяет понять, например, как будет висеть подвесной мост и нужно ли где-то добавлять дополнительные опоры.

Постановка задачи:

1. Положим, что кривая задаётся как $y(x)$, при этом $x \in [0,a]$.
2. Длина кривой (цепи или нити) постоянная, мы полагаем, что она не может растягиваться: 
$$
\int_0^a \sqrt{1+y'^2} dx = L = const.
$$
3. Общая потенциальная энергия считается как сумма потенциальных энергий в каждой точке, иными словами,
$$
E[y] = \int_0^a \rho g y \sqrt{1 + y'^2} dx,
$$
где мы задаём $\rho>0$ -- плотность или массу малого сегмента цепи -- и вспоминаем, что потенциальная энергия тела, поднятого над землёй на высоту $h$, -- это $mgh$ (может быть отрицательная).

Для нас более-менее  понятно, что мы ищем функцию $y \in C^2$, достаточно гладкую кривую, и которая удовлетворяет

*  краевым условиям $y(0)=y_0, y(a)=y_a$ и
*  $\int_0^a \sqrt{1+y'^2} dx = L = const.$

Эти условия задают класс $\mathcal{M}$. Задача:

$$
\min_{y \in \mathcal{M}} E[y].
$$

### Вывод уравнения на экстремали (10 баллов)

$$
где $C$ -- константа, которую надо будет определить. Для краткости заведём $\tilde{\rho}=\rho g$. Чтобы решить задачу, нужно действовать в два этапа: сначала решить уравнение для всех $\lambda, C$, полагая их константами, далее определить $\lambda$ из ограничения, а также $C$ из краевых условий.
